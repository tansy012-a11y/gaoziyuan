## AVL 树构建过程（序列：30, 20, 10, 25, 40, 35, 50）

### 1. 插入 30
空树，创建根结点。
```
30(0)
```
平衡因子：30 左高 0 - 右高 0 = 0
无失衡。

---

### 2. 插入 20
20 < 30，成为左孩子。
```
    30(1)
   /
 20(0)
```
- 结点 20：bf = 0
- 结点 30：左高 1，右高 0 → bf = 1
无失衡。

---

### 3. 插入 10
10 < 30，向左；10 < 20，向左。
```
      30(2)
     /
   20(1)
   /
 10(0)
```
- 结点 10：bf = 0
- 结点 20：左高 1，右高 0 → bf = 1
- 结点 30：左高 2，右高 0 → bf = 2 ⚠️ **失衡**

**失衡类型**：LL（左左）
**旋转轴**：结点 30（右旋）

**右旋后**（以 20 为新根）：
```
    20(0)
   /  \
 10(0) 30(0)
```
所有结点平衡因子为 0。

---

### 4. 插入 25
从根 20 开始：25 > 20 向右，25 < 30 向左。
```
      20(-1)
     /   \
   10(0) 30(1)
        /
     25(0)
```
- 25：bf = 0
- 30：左高 1，右高 0 → bf = 1
- 20：左高 1，右高 2 → 1 - 2 = -1
无失衡（|bf| ≤ 1）。

---

### 5. 插入 40
20 → 30 → 向右。
```
      20(-1)
     /   \
   10(0) 30(0)
        /  \
      25(0) 40(0)
```
- 40：bf = 0
- 30：左高 1，右高 1 → bf = 0
- 20：左高 1，右高 2 → bf = -1
无失衡。

---

### 6. 插入 35
20 → 30 → 40 → 向左（35 < 40）。
```
         20(-2)
        /   \
      10(0) 30(-1)
           /   \
        25(0)  40(1)
               /
            35(0)
```
计算各结点高度及平衡因子：
- 35：bf = 0
- 40：左高 1，右高 0 → bf = 1
- 30：左高 1，右高 2 → bf = -1
- 20：左高 1，右高 3 → bf = -2 ⚠️ **失衡**

**失衡类型**：RL（右左）
**旋转过程**：
① 对结点 40 进行 **右旋**（轴：40）
② 对结点 20 进行 **左旋**（轴：30）

**第一步：右旋 40**（以 35 替代 40 位置）
```
         20
        /   \
      10    30
           /  \
         25    35
                 \
                 40
```

**第二步：左旋 20**（以 30 为新根）
```
           30(0)
          /    \
        20(0)  35(-1)
       /  \       \
     10(0)25(0)   40(0)
```
最终所有结点 |bf| ≤ 1。

---

### 7. 插入 50
从根 30 开始：30 → 35 → 40 → 向右。
```
           30
          /  \
        20    35
       / \     \
     10  25    40
                 \
                 50
```
更新平衡因子（从 50 向上计算）：
- 50：bf = 0
- 40：左空，右高 1 → bf = -1
- 35：左空，右高 2 → bf = -2 ⚠️ **失衡**

**失衡类型**：RR（右右）
**旋转轴**：结点 35（左旋）

**左旋 35**（以 40 为新根）：
```
           30(0)
          /    \
        20(0)  40(0)
       /  \    /  \
     10(0)25(0)35(0)50(0)
```
全部结点平衡因子为 0。

---

## 最终 AVL 树
```
          30
         /  \
        20   40
       / \   / \
      10 25 35 50
```
（每个结点平衡因子均为 0）

## 中序遍历验证
10 → 20 → 25 → 30 → 35 → 40 → 50
**升序排列**，BST 性质保持不变。
