一、数据结构定义

```python
class ListNode:
    def __init__(self, val):
        self.val = val
        self.next = None
```

二、链表反转 reverse()

思路（迭代法，最优解）

通过三个指针逐步反转指向关系：

prev：当前节点的前驱（初始为 None）

curr：当前节点

next_node：暂存后继节点，防止断链

```python
def reverse(head: ListNode) -> ListNode:
    prev = None
    curr = head

    while curr:
        next_node = curr.next  # 保存后继
        curr.next = prev       # 反转指针
        prev = curr            # 前移 prev
        curr = next_node       # 前移 curr

    return prev
```

复杂度
时间复杂度：O(n)
空间复杂度：O(1)

三、环检测 has_cycle()

思路（Floyd 判圈算法 / 快慢指针）

慢指针每次走一步，快指针每次走两步。若存在环，两者必然相遇

```python
def has_cycle(head: ListNode) -> bool:
    slow = head
    fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

        if slow == fast:
            return True

    return False
```

复杂度
时间复杂度：O(n)
空间复杂度：O(1)

五、简单测试示例

```python
# 构造链表 a -> b -> c
a = ListNode('a')
b = ListNode('b')
c = ListNode('c')

a.next = b
b.next = c

# 反转
new_head = reverse(a)

# 输出 c -> b -> a
cur = new_head
while cur:
    print(cur.val, end=" ")
    cur = cur.next

# 构造环
# 构造链表
a = ListNode('a')
b = ListNode('b')
c = ListNode('c')

a.next = b
b.next = c

# 反转
new_head = reverse(a)

# 制造环
a.next = b

print("Has cycle:", has_cycle(new_head))
```

```python
# 构造节点
nodes = [ListNode(chr(ord('a') + i)) for i in range(10)]

# 连接链表
for i in range(len(nodes) - 1):
    nodes[i].next = nodes[i + 1]

head = nodes[0]



new_head = reverse(head)

# 输出反转结果
cur = new_head
while cur:
    print(cur.val, end=" ")
    cur = cur.next


print("\nHas cycle:", has_cycle(new_head))
```
