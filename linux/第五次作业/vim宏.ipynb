**第一阶段：用宏生成 1~255**

Step 1：打开文件
vim result.txt

Step 2：输入第一行
按：
i
输入：
ssh root@10.64.198.1 > 1.log
按：
Esc

Step 3：
录制宏
qq

Step 4：宏内容
① 复制当前行 yy
② 粘贴 p
③ 定位到 IP 最后一个数字
④ 数字 +1 Ctrl + a
⑤ 结束宏 q

Step 5：执行宏
先回到第一行 gg
然后执行 255@q

✔ 这一步完成后得到：
ssh root@10.64.198.1 > 1.log
ssh root@10.64.198.2 > 1.log
ssh root@10.64.198.3 > 1.log
...
ssh root@10.64.198.255 > 1.log
